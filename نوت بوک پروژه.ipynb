    در زیر کد مربوط به جایگزینی مقادیر هر ستون با مقدار مناسب آمده است . همچنین اعداد فارسی را به انگلیسی تبدیل میکنیم تا توسط رایانه درک شود و یا مقادیر بولین را به عددی تبدیل میکنیم تا در گام‌های بعدی بتوان بهتر از آن استفاده کرد و ...

```python
# پاکسازی داده‌ها
df_cleaned = df[df['price_value'].notnull() & df['building_size'].notnull()].copy()
df_cleaned.dropna(subset=['construction_year'], inplace=True)
df_cleaned['construction_year'] = df_cleaned['construction_year'].astype(str)
df_cleaned['construction_year'] = df_cleaned['construction_year'].str.replace('قبل از ۱۳۷۰', '1369', regex=False)

df_cleaned['location_latitude'].fillna(df_cleaned['location_latitude'].median(), inplace=True)
df_cleaned['location_longitude'].fillna(df_cleaned['location_longitude'].median(), inplace=True)
df_cleaned['location_radius'].fillna(df_cleaned['location_radius'].median(), inplace=True)
df_cleaned['created_at_month'] = pd.to_datetime(df_cleaned['created_at_month'], errors='coerce')
```

```python
# تبدیل اعداد فارسی به انگلیسی
def persian_to_english(x):
    if pd.isna(x):
         return None
    x = str(x)
    persian_digits = '۰۱۲۳۴۵۶۷۸۹'
    english_digits = '0123456789'
    return x.translate(str.maketrans(persian_digits, english_digits))

df_cleaned["construction_year"] = df_cleaned["construction_year"].apply(persian_to_english)
df_cleaned['construction_year'] = pd.to_numeric(df_cleaned['construction_year'], errors='coerce').fillna(-1).astype(int)
```

```python
# تبدیل مقادیر باینری به عددی
def convert_to_int_if(value):
    if pd.isnull(value):
        return -1
    else:
        return int(bool(value))

cols = ['has_balcony', 'has_elevator', 'has_warehouse', 'has_parking','is_rebuilt'] 
for col in cols:
    df_cleaned[col] = df_cleaned[col].apply(convert_to_int_if)
```

```python
# ویرایش داده های ستون های مربوط به طبقات
df_cleaned['floor'] = df_cleaned['floor'].apply(persian_to_english)
df_cleaned['floor'] = pd.to_numeric(df_cleaned['floor'], errors='coerce')
df_cleaned['floor'].fillna(df_cleaned['floor'].median(), inplace=True)
df_cleaned['floor'] = df_cleaned['floor'].astype(int)

df_cleaned['total_floors_count'] = df_cleaned['total_floors_count'].apply(persian_to_english)
df_cleaned['total_floors_count'] = pd.to_numeric(df_cleaned['total_floors_count'], errors='coerce')
df_cleaned['total_floors_count'].fillna(df_cleaned['total_floors_count'].median(), inplace=True)
df_cleaned['total_floors_count'] = df_cleaned['total_floors_count'].astype(int)

df_cleaned['unit_per_floor'] = df_cleaned['unit_per_floor'].apply(persian_to_english)
df_cleaned['unit_per_floor'] = pd.to_numeric(df_cleaned['unit_per_floor'], errors='coerce')
df_cleaned['unit_per_floor'].fillna(df_cleaned['unit_per_floor'].median(), inplace=True)
df_cleaned['unit_per_floor'] = df_cleaned['unit_per_floor'].astype(int)
```

```python
# پر کردن مقادیر ناموجود و نامشخص
cols = ['cat2_slug',
        'cat3_slug',
        'city_slug',
        'neighborhood_slug',
        'user_type',
        'deed_type',
        'has_warm_water_provider',
        'has_heating_system',
        'has_cooling_system',
        'has_restroom',
        'building_direction',
        'floor_material']

df_cleaned[cols] = df_cleaned[cols].fillna('unknown')
```

```python
# تبدیل شمارگان اتاق از حروف به عدد
map_dict = {
    'بدون اتاق': 0,
    'یک': 1,
    'دو': 2,
    'سه': 3,
    'چهار': 4,
    'پنج یا بیشتر': 5
}
df_cleaned['rooms_count'] = df_cleaned['rooms_count'].map(map_dict)
df_cleaned['rooms_count'] = df_cleaned['rooms_count'].fillna(-1)
```

```python
# فیلتر کردن
df_clean = df_cleaned[(df_cleaned['building_size'] >= 20 ) & (df_cleaned['building_size'] < 1000)]
df_clean = df_clean[(df_clean['construction_year'] >= 1369) & (df_clean['construction_year'] <= 1404)]
df_clean = df_clean[(df_clean['rooms_count'] <= 20)]
df_clean = df_clean[(df_clean['total_floors_count'] >= 1) & (df_clean['total_floors_count'] <= 100)] 
df_clean = df_clean[(df_clean['floor'] >= -10) & (df_clean['floor'] <= 60)] 
df_clean = df_clean[((df_clean['unit_per_floor'] >= 0) & (df_clean['unit_per_floor'] <= 50))] 
df_clean = df_clean[(df_clean['location_latitude'] >= 25) & (df_clean['location_latitude'] <= 40)]
df_clean = df_clean[(df_clean['location_longitude'] >= 44) & (df_clean['location_longitude'] <= 64)]

# حذف ۱ درصد خیلی ارزان و ۱ درصد خیلی گران
df_clean = df_clean[(df_clean['price_value'] < df_clean['price_value'].quantile(0.99)) & (df_clean['price_value'] > df_clean['price_value'].quantile(0.01))]

# ذخیره داده پاکسازی شده به عنوان فایل داده جدید
df_clean.to_csv('CleanFinall.csv', index=False, encoding='utf-8-sig')
```

```python
# فراخوانی سایر کتابخانه‌های مورد نیاز
import numpy as np
from tqdm import tqdm
from scipy import stats
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import StandardScaler, OneHotEncoder
```

```python
# انتخاب ویژگی‌های عددی فقط
features_numeric = ['price_value', 'building_size', 'is_rebuilt', 'rooms_count', 'floor','construction_year']
```

```python
# پردازش ویژگی‌های عددی
numerical_cols = ['price_value', 'building_size', 'is_rebuilt','rooms_count','floor','construction_year']
df['price_value'] = df['price_value'].fillna(df['price_value'].median())
df['building_size'] =df['building_size'].fillna(df['building_size'].median())
df['is_rebuilt'] =df['is_rebuilt'].fillna(df['is_rebuilt'].median())
df['rooms_count'] = df['rooms_count'].fillna(df['rooms_count'].median())
df['floor'] = df['floor'].fillna(df['floor'].median())
df['construction_year'] = df['construction_year'].fillna(df['construction_year'].median())
```

```python
# مقادیر پرت را شناسایی و پاک میکنیم Z-Score با به کاربردن
z_scores = np.abs(stats.zscore(df[features_numeric].select_dtypes(include=['number'])))
outliers_z = (z_scores > 3)  # بزرگتر از 3 به عنوان مقادیر پرت شناسایی می‌شوند
df_no_outliers = df[~outliers_z.any(axis=1)]  # پاک کردن نمونه‌هایی که مقادیر پرت دارند
```

```python
# نرمالسازی داده عددی
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df_no_outliers[features_numeric])
```

```python
# PCA کاهش ابعاد با
pca = PCA(n_components=2)  # به عنوان نمونه 2 مولفه اصلی
df_pca = pca.fit_transform(df_scaled)
```

```python
inertias = []
sil_scores = []
ks = range(3, 10)
n_sample_for_sil = min(10000,len(df_pca))
indices = np.random.choice(len(df_pca),n_sample_for_sil, replace=False)
df_pca_sample = df_pca[indices]
for k in tqdm(ks,desc="اجرای KMeans روی داده‌های PCA"):
    km = KMeans(n_clusters=k,random_state=42,n_init="auto")
    km.fit(df_pca)
    labels = km.predict(df_pca)
    labels_sample = labels[indices]
    inertias.append(km.inertia_)
    sil_scores.append(silhouette_score(df_pca_sample,labels_sample))
```

```python
# اجرا و مقایسه
print(f"{'Model Name':<25} | {'MAE':<15}      | {'RMSE':<15}      | {'R2 Score':<10}")
print("#" * 80) 
results = {}
for name, model in models.items():
    # آموزش
    model.fit(X_train_scaled, y_train)
    
    # پیش‌بینی
    pred_log = model.predict(X_test_scaled)
    
    # برگرداندن به قیمت واقعی
    pred_real = np.expm1(pred_log)
    y_test_real = np.expm1(y_test)
    
    # محاسبه خطا
    mae = mean_absolute_error(y_test_real, pred_real)
    rmse = math.sqrt(mean_squared_error(y_test_real, pred_real))
    r2 = r2_score(y_test_real, pred_real)
    
    results[name] = r2
       
    print(f"{name:<25} | {mae:,.0f}        | {rmse:,.0f}        | {r2:.4f}")
print("#" * 80)

# یافتن بهترین مدل
best_model = max(results, key=results.get)
print(f"\nبهترین مدل: {best_model} با R2 = {results[best_model]:.4f}")
```

```python
# clean_text در یک ستون نو به نام description و title ترکیب دو ستون
df = df[['title', 'description', 'cat3_slug', 'user_type']].copy()
df_sampled = df.sample(frac=0.1, random_state=42)
df_sampled['clean_text'] = df_sampled['title'] + ' ' + df_sampled['description']
df_sampled.info()
```

```python
# تمیزسازی متن (حذف علائم نگارشی ، اعداد و واژگان توقف)
def clean_text(text):
    text = str(text)
    text = text.lower()                  # تبدیل به حروف کوچک
    text = re.sub(r'[^\w\s]', '', text)  # حذف علائم نگارشی
    text = re.sub(r'\d+', '', text)      # حذف اعداد
    return text
df_sampled['clean_text'] = df_sampled['clean_text'].apply(clean_text)
```

```python
# تبدیل متن به توکن
tokenizer = Tokenizer(num_words=5000)
tokenizer.fit_on_texts(df_sampled["clean_text"])

X_seq = tokenizer.texts_to_sequences(df_sampled["clean_text"])
X_pad = pad_sequences(X_seq, maxlen=80)
```

```python
# کدگذاری برچسب
df_sampled["user_type"] = df_sampled["user_type"].fillna("null")

le_cat = LabelEncoder()
df_sampled["cat3_encoded"] = le_cat.fit_transform(df_sampled["cat3_slug"])

le_user = LabelEncoder()
df_sampled["user_type_encoded"] = le_user.fit_transform(df_sampled["user_type"])

y_cat = df_sampled["cat3_encoded"]
y_user = df_sampled["user_type_encoded"]
```

```python
# تقسیم داده‌ها به داده‌های آموزشی و تست
X_train_text, X_test_text, y_train_cat, y_test_cat = train_test_split(
    df_sampled["clean_text"], y_cat, test_size=0.2, random_state=42)

X_train_pad, X_test_pad, y_train_user, y_test_user = train_test_split(
    X_pad, y_user, test_size=0.2, random_state=42)
```

```python
# BOW و TF-IDF تبدیل متن به بردار با
vectorizer = CountVectorizer(max_features=3000)
tfidf_vectorizer = TfidfVectorizer(max_features=3000)

X_train_bow_vec = vectorizer.fit_transform(X_train_text).toarray()
X_test_bow_vec = vectorizer.transform(X_test_text).toarray()
```

```python
# آموزش مدلهای رگرسیون لجستیک و بیز ساده برای دسته بندی نوع ملک
print("=== آموزش مدل رگرسیون لجستیک ===")
log_reg = LogisticRegression(max_iter=1500)
log_reg.fit(X_train_bow_vec, y_train_cat)
y_pred_log = log_reg.predict(X_test_bow_vec)
print(classification_report(y_test_cat, y_pred_log))
print("#"*50)
print("=== آموزش مدل بیز ساده ===")
nb = MultinomialNB()
nb.fit(X_train_bow_vec, y_train_cat)
y_pred_nb = nb.predict(X_test_bow_vec)
print(classification_report(y_test_cat, y_pred_nb))
```

```python
acc_log = accuracy_score(y_test_cat,y_pred_log)
prec_log = precision_score(y_test_cat, y_pred_log,average="weighted")
rec_log = recall_score(y_test_cat,y_pred_log,average="weighted")
f1_log = f1_score(y_test_cat,y_pred_log,average="weighted")

acc_nb = accuracy_score(y_test_cat,y_pred_nb)
prec_nb = precision_score(y_test_cat, y_pred_nb,average="weighted")
rec_nb = recall_score(y_test_cat,y_pred_nb,average="weighted")
f1_nb = f1_score(y_test_cat,y_pred_nb,average="weighted")

labels = ["Accuracy","Precision","Recall","F1-score"]
log_values = [acc_log,prec_log,rec_log,f1_log]
nb_values = [acc_nb,prec_log,rec_log,f1_log]

x=range(len(labels))

plt.figure(figsize=(10,6))
plt.bar(x,log_values,width=0.4,label="Logistic Regression", color="red")
plt.bar([i+0.4 for i in x],nb_values,width=0.4,label="Naive Byes", color="green")

plt.xlabel("Score")
plt.ylabel("Score")
plt.title("Logistic Regression VS Naive Byes")
plt.xticks([i+0.2 for i in x],labels)
plt.ylim(0,1)
plt.legend()
plt.grid(axis="y",alpha=0.3)

for i , v in enumerate(log_values):
    plt.text(i,v + 0.01,f"{v:.3f}",ha="center",fontweight="bold")
for i,v in enumerate(nb_values):
    plt.text(i+0.4,v+0.01,f"{v:.3f}",ha="center",fontweight="bold")

plt.tight_layout()
plt.show()
```

```python
# آموزش مدل شبکه عصبی برای دسته بندی نوع کاربر
num_classes_user = y_user.nunique()

model_user = tf.keras.Sequential([
    tf.keras.layers.Embedding(input_dim=5000, output_dim=32, input_length=80),
    tf.keras.layers.GlobalAveragePooling1D(),
    tf.keras.layers.Dense(32, activation='relu'),
    tf.keras.layers.Dense(num_classes_user, activation='softmax')])

model_user.compile(optimizer="adam",
                   loss="sparse_categorical_crossentropy",
                   metrics=["accuracy"])

model_user.fit(X_train_pad, y_train_user,
               epochs=3,
               batch_size=128,
               validation_data=(X_test_pad, y_test_user))

user_score = model_user.evaluate(X_test_pad, y_test_user)
print("Neural Network Accuracy (user_type):", user_score[1])
```

در داده‌ی داده شده که به صورت یک دیتافریم پانداس است مقادیر ناموجود زیادی وجود دارد که مزاحم کار ماست . این مقادیر پس از شناسایی یا حذف شدند و یا با مقدار مناسبی جایگزین شدند . سطرهای تکراری و ستونهایی که مقدار NaN بسیاری داشتند نیز حذف شدند تا حجم دیتا فریم کمتر شود و کار با آن بهتر و دقیق تر پیش برود . با توجه به اینکه ما بر روی داده‌های مربوط به خانه‌های مسکونی که برای فروش یا خرید در سایت آگهی شده بودند تمرکز کردیم ، پس ستونها و ردیف هایی که مربوط به این موارد نبودند نیز از دیتافریم پاک شدند . </p> 
<p style="font-family:Nazanin; font-size:150%; text-align:right;" dir="rtl">
   با توجه به ماتریس همبستگی و نمودارهای مربوط به آن دریافتیم که بیشترین تاثیر در افزایش قیمت خانه مربوط به اندازه آن است که رابطه مستقیمی با تعداد اتاق‌های آن دارد . همچنین مواردی مانند سن ساختمان و تعداد واحد موجود در هر طبقه نیز تاثیر منفی در قیمت دارند یعنی هرچه اینچنین مقادیری بیشتر باشند قیمت خانه با کاهش رو به رو می‌شود . 
</p> 
<p style="font-family:Nazanin; font-size:150%; text-align:right;" dir="rtl">
    از این رو میتوان گفت خریداری که میخواهد بهترین خانه که هم ارزش بیشتری دارد و هم قیمت پایین تری دارد را بیابد ، میتواند آنرا در آگهی‌هایی که در آن یک خانه پهناور بازسازی شده در ساختمانی که در هر طبقه تعداد زیادی واحد دیگر نیز وجود دارد را جست و جو کند .
</p> 
<p style="font-family:Nazanin; font-size:150%; text-align:right;" dir="rtl">
    اکنون زمان کاهش بعد و خوشه بندی است که ما با روش PCA کاهش بعد داده و سپس داده را به 3 خوشه تقسیم کردیم و هر خوشه را مورد تفسیر و بررسی قرار دادیم .
</p> 
<p style="font-family:Nazanin; font-size:150%; text-align:right;" dir="rtl">
    پس از آن ، ما به 5 روش رگرسون خطی ، KNN ، جنگل تصادفی ، درخت تصمیم و XGBoost مدلی برای پیش‌بینی قیمت آموزش داده و آنها را باهم مقایسه کردیم و دریافتیم که اگرچه نتایج Random Forest و XGBoost بسیار به هم نزدیک است اما مدل برنده XGBoost است و براساس آن اقدام به پیش‌بینی قیمت خانه و تعیین قیمت منصفانه نسبت به قیمت اعلامی یا پیش‌بینی شده کردیم . بر پایهٰی این پیش‌بینی ها و مدل سازیٰها میتوان گفت که بیشتر خانه‌های مورد بررسی قرار گرفته دارای قیمتی مناسب هستند
</p> 
<p style="font-family:Nazanin; font-size:150%; text-align:right;" dir="rtl">
    همچنین 5 مورد از خانه‌هایی که بیشترین کیفیت و کمترین قیمت را دارند را معرفی کردیم تا خریدار بتواند بر اساس آن تصمیمی بگیرد که به سودش باشد .
</p> 
<p style="font-family:Nazanin; font-size:150%; text-align:right;" dir="rtl">
دو ستون خواسته شده با موفقیت باهم ترکیب و در ستون clean_text ذخیره شدند . متن پاکسازی و تمیز شد و موارد موجود در ستونهای مربوط به نوع کاربر و نوع ملک دسته بندی شده و در پایان مقادیر ناموجود هر یک از این دو ستون پیش بینی شده و در دو ستون cat3_encoded  و user_type_encoded ذخیره شدند . میتوان گفت که مدل رگرسیون لجستیک با دقت 0.79 بهترین مدل برای دسته بندی نوع ملک است و مدل شبکه های عصبی با دقت 0.73 به خوبی نوع کاربر را دسته بندی و پیش بینی کرده است .
</p>
